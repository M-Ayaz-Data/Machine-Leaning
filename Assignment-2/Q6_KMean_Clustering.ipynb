```python
import pandas as pd
from sklearn.cluster import KMeans


data = {
    'X': [1, 2, 4, 5, 8, 9],
    'Y': [2, 1, 5, 4, 9, 8]
}

df = pd.DataFrame(data)


model = KMeans(n_clusters=2, random_state=42, n_init=10)


model.fit(df)

print("Clusters:", model.labels_)
print("Centroids:")
print(model.cluster_centers_)
```
