```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree, export_text

# Step 1: Create dataset
data = {
    'Size_sqft': [1000, 1200, 1500, 1800, 2000],
    'Bedrooms': [2, 2, 3, 3, 4],
    'Price': [50, 60, 80, 90, 120]
}

df = pd.DataFrame(data)

# Step 2: Separate input and output
X = df[['Size_sqft', 'Bedrooms']]
y = df['Price']

# Step 3: Train model
model = DecisionTreeRegressor(max_depth=2, random_state=42)
model.fit(X, y)

# Step 4: Show the learned rules
print(export_text(model, feature_names=list(X.columns)))

# Step 5: Draw the tree
plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=X.columns,
    filled=True,
    rounded=True
)
plt.show()

# Step 6: Predict price for a new house
new = pd.DataFrame([[1600, 3]], columns=X.columns)
pred = model.predict(new)

print("Predicted Price:", pred[0])
```
